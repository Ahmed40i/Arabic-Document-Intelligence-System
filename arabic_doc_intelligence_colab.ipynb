## 4) ارفع المستند (PDF أو صورة) وعالجه

```python
up = files.upload()
path = list(up.keys())[0]
info = pipeline.ingest(path)
print(info)
for p in pipeline.get_pages(info['doc_id']):
    print(f"--- صفحة {p['number']} ({p['source']}) ---")
    print(p['text'][:600])
```

## 5) الاختبار

```python
def ask(q):
    r = pipeline.ask(info['doc_id'], q)
    print('route:', r['route'])
    print(r['answer'] if isinstance(r['answer'], str) else __import__('json').dumps(r['answer'], ensure_ascii=False, indent=1))

ask('لخص لي المستند.')
```

```python
ask('What is the main purpose of this document?')
```
